# 📊 Insights Summary – Retail Sales EDA

This analysis is based on the **Online Retail Dataset**. Below are the key insights derived from the exploratory data analysis:

---

## 1. Customer Behavior
- **Basket Size Distribution**:  
  Most transactions involve **1–3 items**. Bulk purchases are rare but contribute significantly to overall revenue.
  
- **Top Customers**:  
  A small group of customers generate the majority of sales (**Pareto principle: ~20% customers contribute ~80% revenue**).  
  Frequent buyers are not always top revenue contributors — some purchase often but in smaller amounts.

---


## 2. Product Insights
- **Quantity Distribution**:  
  Most items are sold in small quantities, suggesting **retail-oriented behavior**.  
  Outliers (very high quantities) may represent **bulk orders** or **returns**.
  
- **Top Products**:  
  A few products dominate sales, showing **high product concentration**.  
  These products should be prioritized for **stock management, promotions, and cross-selling**.

---

## 3. Revenue Analysis
- **Invoice Revenue Distribution**:  
  Most invoices are low-value, while a few high-value invoices contribute disproportionately to total revenue.  

- **Top Countries**:  
  The **UK accounts for the majority of sales**, while international customers bring smaller but promising contributions.  
  
- **Customer Revenue vs. Frequency**:  
  Some customers purchase less often but spend more per order, making them ideal targets for **premium services/loyalty programs**.

---


## 4. Time-based Insights
- **Monthly Transactions**:  
  Sales peak during **November and December**, reflecting the **holiday shopping season**.  
  → Suggests the need for **seasonal marketing campaigns and inventory ramp-ups**.
  
- **Weekday Transactions**:  
  Transactions are higher on **weekdays (Tue–Thu)** compared to weekends, indicating **B2B or retail reseller purchases**.
